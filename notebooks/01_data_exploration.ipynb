```python
import pandas as pd
import os

# Tạo thư mục processed nếu chưa có
os.makedirs("data/processed", exist_ok=True)

# 1. Đọc dữ liệu thô từ thư mục raw
print("Đang đọc dữ liệu gốc...")
train_raw = pd.read_csv("../data/raw/train.csv")
test_raw = pd.read_csv("../data/raw/test.csv")

# 2. Làm sạch và gộp cột cho tập Train
df_clean_train = pd.DataFrame()
df_clean_train['label'] = train_raw['Class Index']
df_clean_train['text'] = train_raw['Title'] + ". " + train_raw['Description']

# 3. Làm sạch và gộp cột cho tập Test
df_clean_test = pd.DataFrame()
df_clean_test['label'] = test_raw['Class Index']
df_clean_test['text'] = test_raw['Title'] + ". " + test_raw['Description']

# 4. Lấy mẫu dữ liệu để thử nghiệm thuật toán (1.000 dòng)
df_clean = df_clean_train

# 5. Lưu kết quả vào thư mục processed
df_clean.to_csv("../data/processed/df_clean.csv", index=False)
df_clean_test.to_csv("../data/processed/test_clean.csv", index=False)

print("\nĐã lưu thành công dữ liệu vào data/processed/")
print("Kích thước df_clean hiện tại:", df_clean.shape)
print("\nXem thử 3 dòng đầu tiên:")
print(df_clean.head(3))
```
